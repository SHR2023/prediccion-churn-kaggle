```r
library(readxl)
library(tidyverse)
library(caret)
library(pROC)
library(xgboost)
```

```r
#cargar datos
train <- read_excel("traintelco.xlsx")
test <- read_excel("testelco.xlsx")

train_id <- train$id
test_id <- test$id
```

```r
#ingenieria de variables
feature_engineering <- function(df){

  fecha_corte <- as.POSIXct("2019-01-01")

  df$edad <-
    as.numeric(
      difftime(
        fecha_corte,
        df$`Fecha de nacimiento`,
        units = "days"
      )
    ) / 365.25

  df$antiguedad_cliente <-
    as.numeric(
      difftime(
        fecha_corte,
        df$`Fecha inicio contrato`,
        units = "days"
      )
    )

  df$facturacion_mes <-
    df$facturación /
    (df$antiguedad_cliente/30 + 1)

  df$minutos_mes <-
    df$minutos /
    (df$antiguedad_cliente/30 + 1)

  df$facturacion_por_minuto <-
    df$facturación /
    (df$minutos + 1)

  df$mora_antiguedad <-
    df$mora /
    (df$antiguedad_cliente + 1)

  df$facturacion_antiguedad <-
    df$facturación /
    (df$antiguedad_cliente + 1)

  df$log_facturacion <-
    sign(df$facturación) *
    log1p(abs(df$facturación))

  df$log_minutos <-
    log1p(df$minutos)

  df$log_mora <-
    log1p(df$mora)

  df$mora_facturacion <-
    df$mora /
    (abs(df$facturación) + 1)

  df$mora_minutos <-
    df$mora /
    (df$minutos + 1)

  df$equipo_cliente_ratio <-
    df$`Antigüedad Equipo` /
    (df$antiguedad_cliente + 1)

  # NUEVAS VARIABLES

  df$edad_x_facturacion <-
    df$edad * df$facturación

  df$edad_x_minutos <-
    df$edad * df$minutos

  df$facturacion_x_minutos <-
    df$facturación * df$minutos

  df$facturacion_x_antiguedad <-
    df$facturación * df$antiguedad_cliente

  df$edad2 <-
    df$edad^2

  df$minutos2 <-
    df$minutos^2

  df$facturacion2 <-
    df$facturación^2

  return(df)
}

train <- feature_engineering(train)
test <- feature_engineering(test)
colSums(is.na(train))
```

```r
#ver columnas vacias
colSums(is.na(train))
```

```r
#convertir a cetegòricas
train$`tipo cliente` <- as.factor(train$`tipo cliente`)
train$`Factura online` <- as.factor(train$`Factura online`)
train$`Plan de datos` <- as.factor(train$`Plan de datos`)

test$`tipo cliente` <- as.factor(test$`tipo cliente`)
test$`Factura online` <- as.factor(test$`Factura online`)
test$`Plan de datos` <- as.factor(test$`Plan de datos`)
```

```r
#ver tipos
table(train$`tipo cliente`)
table(train$`Factura online`)
table(train$`Plan de datos`)
```

```r
#eliminar columnas que genern fugas o problemas
drop_cols <- c(
  "id",
  "Fecha de nacimiento",
  "Fecha inicio contrato"
)
```

```r
#separar x e y
y <- train$resultado

X <- train %>%
  select(-resultado) %>%
  select(-all_of(drop_cols))

X_test <- test %>%
  select(-all_of(drop_cols))
```

```r
#one hot encoding
dummies <- dummyVars(
  ~ .,
  data = X
)

X <- predict(
  dummies,
  X
) %>% as.data.frame()

X_test <- predict(
  dummies,
  X_test
) %>% as.data.frame()

```

```r
#validaciòn cruzada
set.seed(123)

folds <- createFolds(
  y,
  k = 5,
  returnTrain = FALSE
)
```

```r
#xgboostA
oofA <- rep(0, nrow(X))
predA <- rep(0, nrow(X_test))

for(i in 1:5){

  cat("\nFold:",i,"\n")

  valid_idx <- folds[[i]]

  train_idx <- setdiff(
    1:nrow(X),
    valid_idx
  )

  dtrain <- xgb.DMatrix(
    data = as.matrix(X[train_idx,]),
    label = y[train_idx]
  )

  dvalid <- xgb.DMatrix(
    data = as.matrix(X[valid_idx,]),
    label = y[valid_idx]
  )

  dtest <- xgb.DMatrix(
    data = as.matrix(X_test)
  )

  model <- xgb.train(

    params = list(
      objective = "binary:logistic",
      eval_metric = "auc",

      eta = 0.01,
      max_depth = 4,
      subsample = 0.8,
      colsample_bytree = 0.8,
      min_child_weight = 5,
      gamma = 0.5
    ),

    data = dtrain,

    nrounds = 4000,

    watchlist = list(
      train = dtrain,
      valid = dvalid
    ),

    early_stopping_rounds = 200,

    verbose = 0
  )

  oofA[valid_idx] <-
    predict(
      model,
      dvalid
    )

  predA <-
    predA +
    predict(
      model,
      dtest
    ) / 5
}
```

```r
#XgboostB
oofB <- rep(0, nrow(X))
predB <- rep(0, nrow(X_test))

for(i in 1:5){

  cat("\nFold:",i,"\n")

  valid_idx <- folds[[i]]

  train_idx <- setdiff(
    1:nrow(X),
    valid_idx
  )

  dtrain <- xgb.DMatrix(
    data = as.matrix(X[train_idx,]),
    label = y[train_idx]
  )

  dvalid <- xgb.DMatrix(
    data = as.matrix(X[valid_idx,]),
    label = y[valid_idx]
  )

  dtest <- xgb.DMatrix(
    data = as.matrix(X_test)
  )

  model <- xgb.train(

    params = list(
      objective = "binary:logistic",
      eval_metric = "auc",

      eta = 0.03,
      max_depth = 3,
      subsample = 0.9,
      colsample_bytree = 0.9,
      min_child_weight = 8,
      gamma = 1
    ),

    data = dtrain,

    nrounds = 4000,

    watchlist = list(
      train = dtrain,
      valid = dvalid
    ),

    early_stopping_rounds = 200,

    verbose = 0
  )

  oofB[valid_idx] <- predict(model, dvalid)

  predB <- predB +
    predict(model, dtest)/5
}


```

```r
#auc xgboost
aucA <- roc(y, oofA)$auc
aucB <- roc(y, oofB)$auc

aucA
aucB
```

```r
#ensamble
oof_final <- 0.7 * oofA +
  0.3 * oofB

pred_final <- 0.7 * predA +
  0.3 * predB
```

```r
#auc ensamble
auc_ensemble <- roc(
  y,
  oof_final
)$auc

auc_ensemble
```

```r
#crear archivo para competencia
submission <- data.frame(
  id = test_id,
  resultado = pred_final
)

write.csv(
  submission,
  "submission5.csv",
  row.names = FALSE
)
```
